```julia
using LinearAlgebra
using Base
```

```julia
dot(step(range(0,1,length=100)) * [.5; ones(length(range(0,1,length=100))-2);.5], range(0,1,length=100).^2)
```

```julia
struct Weights
    r::StepRangeLen
end

Base.iterate(W::Weights, state=1) = state > length(W.r) ? nothing : ((step(W.r) * [.5 ones(length(W.r)-2)... .5])[state], state+1)
Base.size(W::Weights) = length(W.r)
Base.IndexStyle(::Type{<:StepRangeLen}) = IndexLinear()
Base.getindex(W::Weights, i::Int) = W.r[i]
```

```julia
dot(Weights(range(0,1,length=100)), range(0,1,length=100).^2)
```
